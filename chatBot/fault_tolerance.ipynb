```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from  langgraph.checkpoint.memory import InMemorySaver
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq
import time
```

```python
load_dotenv()
```

```python
model=ChatGroq(model="openai/gpt-oss-120b",temperature=0.2,api_key=os.getenv("GROQ_API_KEY"))
```

```python
class fault(TypedDict):
    input:str
    step1:str
    step2:str
    step3:str
```

```python
def step1(state:fault)->fault:
    print("step 1 excuted")
    return {"step1":"done","input":state["input"]}
```

```python
def step2(state:fault)->fault:
    print("step 2")
    time.sleep(30)
    return{"step2":"done"}
    
```

```python
def step3(state:fault)->fault:
    print("step 3")
    return{"step3":"done"}
```

```python
graph=StateGraph(fault)
graph.add_node("step1",step1)
graph.add_node("step2",step2)
graph.add_node("step3",step3)

graph.add_edge(START,"step1")
graph.add_edge("step1","step2")
graph.add_edge("step2","step3")

checkpointer=InMemorySaver()

workflow=graph.compiler(checkpointer=checkpointer)
```

```python
workflow
```

```python
try:
    config1={"configurable":{"thread_id":"1"}}
    workflow.invoke({"input":"start"}, config=config1)
except KeyboardInterrupt:
    print("mannully interupt")
```
